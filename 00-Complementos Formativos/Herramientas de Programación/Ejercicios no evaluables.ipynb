```python
def media_columnas(lst2):
    c=[0]*len(lst2[0])   #Se usa el [0] para referirse a las columnas en vez de a las filas
    for fila in lst2:
        i=0
        for valor in fila:
            c[i]+=valor/len(lst2)
            i+=1
    return c
```

```python
def es_diagonal(m):
    n=len(m)
    for i in range(n):   #range va del 0 hasta el n sin incluir el n
        for j in range(n):
            if i != j and m[i][j]!=0:
                return False
    return True
```

```python
es_diagonal([[1,0,0],[0,2,0],[0,0,3]])
```

```python
def es_identidad(m): 
    n=len(m)
    for i in range(n):
        for j in range(n):
            if (i != j and m[i][j]!=0) or (i==j and m[i][j]!=1):
                return False
    return True
```

```python
es_identidad([[1,0,0],[0,1,0],[0,0,1]])
```

```python
def vip(lst:list):
    c=[]
    for comprador in lst:
        if sum(comprador[1:])>1000 and len(comprador[1:])>=3:
            c.append(comprador[0])
    return sorted(c) #ordena alfabéticamente una lista o de menor a mayor, si se usa sorted(c, reverse=True) es de mayor a menor
```

```python
def comienza_letra(s,letra):
    contador=0
    for palabra in s.split(): #split divide el texto en palabras
        if palabra[0].lower()==letra.lower():
            contador+=1
    return contador
```

```python
comienza_letra('Este día comenzamos a estudiar algoritmos', 'e')
```

```python
def palabra_mas_larga(s):
    i=0
    for palabra in s.split(): #Al igual que antes, separamos el texto por palabras para comparar la longitud de estas y obtener el máximo
        if i==0:
            larga=palabra
            i=1
        if len(palabra)>len(larga):
            larga=palabra
    return larga
```

```python
palabra_mas_larga('Ayer comenzamos a estudiar algoritmos')
```

```python
def Mayus(str):
    return str[0].isupper()

def BuscaTexto(lst):
    contador=0
    for palabra in lst:
        if Mayus(palabra)==True:
            contador+=1
    if contador>=2:
        return True
    else:
        return False
        
```

```python
def PositNegat(lst):
    pos=[]
    neg=[]
    for num in lst:
        if num>=0:
            pos.append(num)
        else:
            neg.append(num)
    return pos, neg
```
