## 01- 1.25pt
- Dado un diccionario con una lista de nombres de películas y al año concatenado, crear una función que devuelva un DataFrame con dos columnas:
  - Columna 1, el nombre de la película sin el año
  - Columna 2, el año de la película

```python
# Your solution HERE !!!!
def extract_year(peliculas_año):
    name=[]
    year=[]
    for peliculas in peliculas_año.values():
        for pelicula in peliculas:
            pel_comp=pelicula.split(" - ")
            name.append(pel_comp[0].strip())
            year.append(pel_comp[1].strip())
        diccionario={"Name": name, "Year": year}
        diccionario=pd.DataFrame(diccionario)
        return diccionario
```

```python
# Test Use Case
source_movies = {'Name': ["Matrix - 1999", "Cadena Perpetua - 1994", "Parasitos - 2019", "Vengadores, Endgame - 2019", "Interstellar - 2014", 
                   "Django Desencadenado - 2012", "El Padrino - 1972", "El Caballero Oscuro - 2008", "Erase Una Vez En America - 1984", 
                   "El Senyor de los Anillos, La Comunidad del Anillo - 2001"]}

expected_output = pd.DataFrame({'Name':["Matrix", "Cadena Perpetua", "Parasitos", "Vengadores, Endgame", "Interstellar", "Django Desencadenado", 
                                        "El Padrino", "El Caballero Oscuro", "Erase Una Vez En America", "El Senyor de los Anillos, La Comunidad del Anillo"],
                                'Year':["1999", "1994", "2019", "2019", "2014", "2012", "1972", "2008", "1984", "2001"]})

output = extract_year(source_movies)   # This will call your function extract_year

display(expected_output)
display(output)

assert(expected_output.equals(output))  # This will fail if the result is not as expected
```

## 02 - 1.25 pt

- Crear una matriz cuadrada 6x6 con valores aleatorios enteros en el rango [-10,10]
- Restar a cada elemento de dicha matriz la media de su correspondiente columna (dicha media redondeada hacia el entero superior más cercano)
- Sustituir los valores negativos múltiplos de 3 por Nan
- Eliminar aquellas columnas que tengan 2 o más celdas con valores Nan

```python
# Your solution HERE !!!!

# Crear una matriz cuadrada 6x6 con valores aleatorios enteros en el rango [-10,10]
a=np.random.randint(low=-10, high=11, size=(6, 6))
print(a)

# Restar a cada elemento de dicha matriz la media de su correspondiente columna (dicha media redondeada hacia el entero superior más cercano)
z=np.shape(a)[1]
for ind_col in range(0, z):
    media=np.mean(a[:, ind_col])
    media=np.ceil(media)
    a[:, ind_col]=a[:, ind_col]-media

# Sustituir los valores negativos múltiplos de 3 por Nan
frame = pd.DataFrame(a)
display(frame)

frame_nan = frame.mask((frame % 3 == 0) & (frame < 0))
display(frame_nan)

# Eliminar aquellas columnas que tengan 2 o más celdas con valores Nan
frame_nan=frame_nan.dropna(thresh=z-1, axis=1)
display(frame_nan)
```

- Dado un array de elementos, escribir una función, que reciba dicho array como parámetro de entrada y devuelva el elemento repetido. Usando plain Python, sin ninguna librería.

```python
# Your solution HERE !!!!
def repeated_element(array):
    no_repetido=[]
    repetido=[]
    for n in array: 
        if n not in no_repetido: 
            no_repetido.append(n)
        else: 
            return n
    return "No hay ningún repetido"
```

```python
# Test Use Case
source_array = [4, 8, 0, 3, 6, 2, 1, 5, 3, 9, 7]
expected_rep_element = 3

rep_element = repeated_element(source_array) # This calls your function repeated_element

print(rep_element)

assert(expected_rep_element == rep_element)  # This will fail if the element is not found correctly
```

```python
# Se establece la ruta del dataset
ruta = os.path.join("res", "aircraft.csv")
df_aircraft = pd.read_csv(ruta, sep=",")
```

```python
# 1.A. Muestra las seis primeras filas.
df_aircraft.head(6)
```

```python
# 1.B. Calcula el porcentaje de valores nulos por columna.
df_aircraft.isnull().sum()/len(df_aircraft)*100
```

```python
# 1.C. Descarta del df original los registros cuyos valores de latitud y longitud no se encuentren en el rango 
# de coordenadas esféricas, siendo la longitud valida en el rango [-180, 180], y la latitud valida en el rango [-90, 90]

df_aircraft = df_aircraft[(df_aircraft['LATITUDE'] >= -90) & (df_aircraft['LATITUDE'] <= 90) & (df_aircraft['LONGITUDE'] >= -180) & (df_aircraft['LONGITUDE'] <= 180)]

df_aircraft
```

```python
# 1.D. Muestra en una figura, con un tamaño de 8 de ancho y 4 de alto, una gráfica de tipo puntos (scatter),
# la localización de los incidentes en base a la longitud y latitud (ya limpios los rangos invalidos en el punto anterior).

fig, ax = plt.subplots(figsize=(8, 4))

y=df_aircraft['LATITUDE']
x=df_aircraft['LONGITUDE']

ax.scatter(x, y, c="red" , label="Accidente" , alpha=0.4, s=6, linewidth=0.4, edgecolor='black')
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.title("Coordenadas accidentes aéreos")


ax.legend()
plt.show()
```

```python
# 1.E. Muestra en una figura un conteo de incidentes sobre la columna 'AIRCRAFT' por año, en el gráfico debe aparecer
# una columna por cada 'DAMAGE_LEVEL' distinto en el dataframe. La gráfica debe ser de tipo linea con diferentes subplots.
# Pista, agrupa primero y aplica una gráfica de relación.



# Se normaliza la fecha
df_aircraft["INCIDENT_DATE"] = pd.to_datetime(df_aircraft["INCIDENT_DATE"], errors="coerce")

# Se añade una columna con solo los años de los accidentes
df_aircraft["YEAR"] = df_aircraft["INCIDENT_DATE"].dt.year

# Agrupamos
grouped = df_aircraft.groupby(["YEAR", "AIRCRAFT", "DAMAGE_LEVEL"]).size().reset_index(name="COUNT")

# Obtenemos el número de estados y los estados de damage_level
estados=grouped["DAMAGE_LEVEL"].unique()
print(len(estados))
print(estados)


```

```python
# Paso 1. Crear figura y varios ejes
fig, axs = plt.subplots(1, len(estados), figsize=(18, 6)) # Iterar por cada nivel de daño y graficar

for ax, estado in zip(axs, estados): 
    subplot = grouped[grouped["DAMAGE_LEVEL"] == estado] 
    
    for aircraft in subplot["AIRCRAFT"].unique(): 
        aircraft_data = subplot[subplot["AIRCRAFT"] == aircraft] 
        ax.plot(aircraft_data["YEAR"], aircraft_data["COUNT"]) 
        ax.set_title(f"Damage Level: {estado}") 
        ax.set_ylabel("Incident Count") 
        ax.set_xlabel("Year")


# Paso 5: ajustes globales
plt.tight_layout () # Evita solapamientos

# Paso 6: mostrar el gráfico
plt.show ()
```

```python
# 2.B. Cambia los valores a minúsculas de columna de nombre de aeropuerto en este dataframe y en el del apartado anterior.
def minus(celda):
    return celda.lower().strip()


df_airports["name"]=df_airports["name"].apply(minus)
df_aircraft['AIRPORT']=df_aircraft['AIRPORT'].apply(minus)
```

```python
display(df_airports["name"]) 
display(df_aircraft['AIRPORT'])
```

```python
# 2.C. Con el propósito de disponer del país de los aeropuertos, junta ambos dataframes para obtener un nuevo dataframe
# con todos los registros de incidentes de avión y los países de los aeropuertos. Como clave, utilizar el nombre del
# aeropuerto modificado en el anterior apartado.

df_merge=pd.merge(df_aircraft , df_airports , left_on ="AIRPORT", right_on ="name", how="left")
```

```python
# 2.D. Como resultado se debe disponer un dataframe con 254222 registros.
df_merge.shape
```

```python
# 2.E. Muestra el número de registros que no disponen de dato en el atributo "iso_country".
df_merge["iso_country"].isnull().sum()
```

```python
# 2.F. Elimina todos los registros que que tengan nulos alguna de las columnas 'DAMAGE_LEVEL' y 'iso_country'.
df_merge = df_merge.dropna(subset=["iso_country", "DAMAGE_LEVEL"]).reset_index(drop=True, inplace=False)
df_merge
```

```python
# 2.G. Sobre el resultado anterior, crea una tabla contando el número de incidentes por país, en las filas deben
# aparecer todos los paises posibles y en las columnas debe de estar el tipo de 'DAMAGE_LEVEL'.

tabla = pd.pivot_table(df_merge, index="iso_country", columns="DAMAGE_LEVEL", values="AIRCRAFT", aggfunc="count")
tabla.fillna(0)
```
