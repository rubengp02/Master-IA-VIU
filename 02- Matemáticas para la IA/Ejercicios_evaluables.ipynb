a) [0.5 punto] Implementa una función, determinante_recursivo, que obtenga el determinante de una matriz cuadrada utilizando la definición recursiva de Laplace

```python
def determinante_recursivo(A):
    if len(A)==1:
        return A[0][0]
    if len(A)==2:
        return A[0][0]*A[1][1]-A[0][1]*A[1][0]
    det=0

    for j in range(len(A)):
        signo=(-1)**j
        submatriz=[fila[:j]+fila[j+1:] for fila in A[1:]]
        det+=signo*A[0][j]*determinante_recursivo(submatriz)

    return det
```

```python
A = [[1, 2, 3],[4, 5, 6],[7, 8, 9]]
print(determinante_recursivo(A))
A = [[2, 1, 3],[0, -1, 4],[5, 2, 0]]
print(determinante_recursivo(A))
```

El método de eliminación de Gauss con pivoteo parcial usa las operaciones mencionadas en el apartado anterior: intercambio de fila por otra fila y la suma de fila con otra fila. Este método consiste en seleccionar la fila que tiene el mayor valor absoluto columna a columna y se intercambia con la fila actual, para luego realizar la suma/resta correspondiente de la nueva fila que contiene el pivote que nos interesa, para eliminar debajo del pivote, creando así una matriz triangular superior. A continuación vemos un ejemplo:

```python
A=[[3,-1,2],
   [5,4,1],
   [1,-3,5]]

A=[[5,4,1], #cambiamos fila 2 por fila 1, ya que 5>3
   [3,-1,2],
   [1,-3,5]]

A=[[5,4,1], 
   [0, -17/5, 7/5], #eliminamos debajo del pivote mediante combinaciones lineales F2-3*F1/5
   [0, 19/5, 24/5]] #F3-F1/5

A=[[5,4,1], 
   [0, 19/5, 24/5],  #cambiamos F2 por F3 ya que 19/5>17/5.
   [0, 0, -55/19]]  #F3-17*F2/19
```

e) [0.5 puntos] ¿Cómo se podría calcular el determinante de una matriz haciendo beneficio de la estrategia anterior y del efecto de aplicar las operaciones elementales pertinentes? Implementa una nueva función, determinante_gauss, que calcule el determinante de una matriz utilizando eliminación gaussiana.


Al realizar la eliminación gaussiana, el cálculo del determinante se vuelve mucho más sencillo, pues únicamente tenemos que multiplicar los elementos de la diagonal

```python
def determinante_gauss(A):
    n = len(A)
    cambios = 0   # Lleva la cuenta de intercambios de fila

    for k in range(n - 1):  # Columna y fila del pivote
        
        # Creamos una lista de posibles filas que pueden tener el mayor pivote
        filas_candidatas = list(range(k, n))

        # Lista de los valores absoluto de la columna k
        valor_absoluto = [abs(A[i][k]) for i in filas_candidatas]

        # Mayor valor absoluto
        valor_max = max(valor_absoluto)

        # Índice del mayor valor absoluto dentro de la lista anteriormente creada
        indice_local = valor_absoluto.index(valor_max)

        # Convertimos a índice real de fila
        fila_max = filas_candidatas[indice_local]

        
        if fila_max != k:  # Intercambiamos filas si es necesario
            A[k], A[fila_max] = A[fila_max], A[k]
            cambios += 1

        # Si el pivote es 0 el determinante será 0
        if A[k][k] == 0:
            return 0

        # Eliminación hacia abajo
        for i in range(k + 1, n): # Tomamos las filas por debajo de la fila del pivote (k)
            m = A[i][k] / A[k][k]  # Coeficiente de cancelación
            for j in range(k, n):   # Actualizamos las columnas
                A[i][j] -= m * A[k][j]

  
    det = 1
    for i in range(n):
        det *= A[i][i]   # Producto de la diagonal


    if cambios % 2 == 1: # Signo negativo si hemos realizado cambios de fila impares
        det = -det

    return det
```

```python
A = [[2, 1, 3],[0, -1, 4],[5, 2, 0]]
print(determinante_gauss(A))   # Comparando con el valor que obtenemos de la función del apartado a), podemos comprobar que es correcto
```
